**[2.1]**

Write a function in the cell below that loads in one of the sets of images and computes their average. You can use the **os.listdir** to get the list of files in the directory. As you load in the images, you should compute an average image on the fly (i.e., keep track of the cumulative average). Don't load all the images at once and then average them! Color images are represented by a 3-dimensional array of size (HxWx3) where the third dimension indexes the red, green and blue channels. 

You should encapsulate your code in a function called **average_image** that takes the image directory as an input and returns the average of the images in that directory. Your function should implement some error checking. Specifically your function should skip over any files in the directory that are not images (**plt.imread** will thrown an **OSError** if the file is not an image). It should ignore images that are not color images. Finally, it should also skip any images which are not the same height and width as the first color image you load in.


````python
#
# these are the only modules needed for problem #2
#
import numpy as np
import os as os
import matplotlib.pyplot as plt
````

````python
def average_image(dirname):
    """
    Computes the average of all color images in a specified directory and returns the result.
    
    The function ignores any images that are not color images and ignores any images that are not
    the same size as the first color image you load in.
    
    Parameters
    ----------
    dirname : str
        Directory to search for images.
        
    Returns
    -------
    numpy.array (dtype=float)
        HxWx3 array containing the average of the images found.
        
    """
    cumulative_sum = None
    count = 0
    image_shape = None

    for file in os.listdir(dirname):
        filepath = os.path.join(dirname, file)
        try:
            img = plt.imread(filepath)
            # Check if it's a color image
            if img.ndim == 3 and img.shape[2] == 3:
                if cumulative_sum is None:
                    # Initialize with the shape of the first valid image
                    cumulative_sum = np.zeros_like(img, dtype=np.float64)
                    image_shape = img.shape
                elif img.shape != image_shape:
                    # Skip images with a different shape
                    continue
                # Add image to the cumulative sum
                cumulative_sum += img
                count += 1
        except (OSError, ValueError):
            # Skip files that are not valid images
            continue

    if count > 0:
        return cumulative_sum / count  # Average the images
    else:
        raise ValueError("No valid color images found in the directory.")
````

**[2.2]**

Write code below which calls your **average_image()** function twice, once for each set of images. Display the resulting average images. Also display a single example image from each set for comparison


````python
set1_dir = "assinment 1/set1"
set2_dir = "assinment 1/set2"

# First, display an example image and then the average image
def display_example_and_average(dirname):
    files = os.listdir(dirname)
    example_image = None

    # Find and display an example image
    for file in files:
        filepath = os.path.join(dirname, file)
        try:
            example_image = plt.imread(filepath)
            if example_image.ndim == 3 and example_image.shape[2] == 3:
                break
        except:
            continue

    if example_image is not None:
        plt.figure(figsize=(10, 5))
        plt.subplot(1, 2, 1)
        plt.imshow(example_image / 255.0)  # Normalize for display
        plt.title("Example Image")
        plt.axis("off")

        # Compute and display the average image
        avg_image = average_image(dirname)
        plt.subplot(1, 2, 2)
        plt.imshow(avg_image / 255.0)  # Normalize for display
        plt.title("Average Image")
        plt.axis("off")
        plt.show()
    else:
        print(f"No valid color images found in {dirname}")

# Display results for both sets of images
print("Displaying results for set1:")
display_example_and_average(set1_dir)

print("Displaying results for set2:")
display_example_and_average(set2_dir)
````

Discuss the appearance of the average images.  Give an explanation as to why the average image
does not look like the individual example image.

The average images will in most cases be smoother, with more blur compared to the individual example images. They can also have more subdued colors, since the process of averaging combines the pixel values from different images. Colors which vary greatly between images turn into pastel shades in the mean image. In other words, averaging red and blue regions across images results in hues closer to purple, hence leading to a decrease in saturation. If images are not perfectly aligned or taken of different subjects, then features such as faces, objects, or landscapes may appear only as ghostly outlines or unclear shapes in the mean image. What happens is that corresponding pixels from distinct images do not align perfectly and thus partially superimpose each other or even fade away.
